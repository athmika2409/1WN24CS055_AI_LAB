```python
def calculate_cost(state):
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def get_best_neighbor(state):
    n = len(state)
    best_state = None
    best_cost = float('inf')
    best_swap = None
    neighbors = []

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            cost = calculate_cost(neighbor)
            neighbors.append(((i, j), tuple(neighbor), cost))

            if cost < best_cost:
                best_cost = cost
                best_state = tuple(neighbor)
                best_swap = (i, j)

    return best_state, best_cost, best_swap, neighbors

def hill_climbing_nqueens(initial_state):
    current_state = tuple(initial_state)
    current_cost = calculate_cost(current_state)
    step = 0

    print(f"Initial State: {current_state} | Cost: {current_cost}\n" + "-"*45)

    while True:
        step += 1
        best_neighbor, best_cost, best_swap, neighbors = get_best_neighbor(current_state)

        print(f"\n--- STEP {step} ---")
        print(f"Current State: {current_state} (Cost = {current_cost})")
        print("Neighbor Evaluations:")
        for swap, n_state, cost in neighbors:
            print(f"  Swap (x{swap[0]}, x{swap[1]}) -> State {n_state} | Cost = {cost}")

        if best_cost < current_cost:
            print(f"\n=> Transitioning to best neighbor: {best_neighbor} (Cost = {best_cost}) via Swap (x{best_swap[0]}, x{best_swap[1]})")
            current_state = best_neighbor
            current_cost = best_cost
        else:
            print("\n=> Terminated: No strictly better neighbor found.")
            break

        if current_cost == 0:
            print(f"\n*** Goal State Reached: {current_state} with Cost = 0 ***")
            break

initial_board = (3, 1, 2, 0)
hill_climbing_nqueens(initial_board)
print("1WN24CS055-ATHMIKA")
```
