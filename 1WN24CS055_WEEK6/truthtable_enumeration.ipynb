```python
import itertools


def implies(a, b):
    return (not a) or b


def evaluate_kb(p, q, r):
    s1 = implies(q, p)
    s2 = implies(p, not q)
    s3 = q or r
    kb = s1 and s2 and s3
    return s1, s2, s3, kb


def truth_table_enumeration():
    symbols = ['P', 'Q', 'R']
    models = list(itertools.product([True, False], repeat=3))

    print(f"{'P':<6}{'Q':<6}{'R':<6}| {'Q->P':<8}{'P->~Q':<8}{'QvR':<8}| {'KB':<8}")
    print("-" * 55)

    kb_true_models = []

    for p, q, r in models:
        s1, s2, s3, kb = evaluate_kb(p, q, r)
        model_dict = {'P': p, 'Q': q, 'R': r}

        if kb:
            kb_true_models.append(model_dict)

        print(
            f"{str(p):<6}{str(q):<6}{str(r):<6}| {str(s1):<8}{str(s2):<8}{str(s3):<8}| {str(kb):<8}"
        )

    print("\nModels in which KB is True:")
    for m in kb_true_models:
        print(m)

    return kb_true_models


def check_entailment(kb_models, query_func, query_name):
    entails = all(query_func(m['P'], m['Q'], m['R']) for m in kb_models)
    print(f"\nDoes KB entail {query_name}? {'YES' if entails else 'NO'}")
    return entails


kb_models = truth_table_enumeration()

check_entailment(kb_models, lambda p, q, r: r, "R")
check_entailment(kb_models, lambda p, q, r: implies(r, p), "R -> P")
check_entailment(kb_models, lambda p, q, r: implies(q, r), "Q -> R")
```
